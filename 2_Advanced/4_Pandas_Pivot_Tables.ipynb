```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset

import matplotlib.pyplot as plt

# Loading the dataset
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset['train'].to_pandas()

#data cleaning
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

Counts of Job Titles

```python
df.pivot_table(index='job_title_short', aggfunc='size')
```

```python
df.groupby('job_title_short').size()
```

```python
df.pivot_table(values='salary_year_avg', index='job_title_short', aggfunc='median')
```

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

```python
df.pivot_table(values='salary_year_avg', index='job_country', columns='job_title_short', aggfunc='median')
```

```python
top_countries = df['job_country'].value_counts().head(5).index

df_top_salary = df.pivot_table(values='salary_year_avg', index='job_country', columns='job_title_short', aggfunc='median')

df_job_country_salary = df_top_salary.loc[top_countries]

df_job_country_salary
```

```python
top_countries = df['job_country'].value_counts().head(5).index

df_top_salary = df.pivot_table(values='salary_year_avg', index='job_country', columns='job_title_short', aggfunc='median')

df_job_country_salary = df_top_salary.loc[top_countries]

job_titles= ['Data Scientist', 'Data Engineer', 'Data Analyst', 'Machine Learning Engineer']

df_job_country_salary = df_job_country_salary[job_titles]

df_job_country_salary.plot(kind='bar')
plt.ylabel('Median Salary (USD)')
plt.title('Median Salary by Job Title and Country')
plt.xticks(rotation=45, ha='right')
plt.show()
```
