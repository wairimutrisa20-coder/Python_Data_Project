CALCULATE PROJECT SALARY NEXT YEAR  apply()

```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset

import matplotlib.pyplot as plt

# Loading the dataset
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset['train'].to_pandas()

#data cleaning
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df[pd.notna(df['salary_year_avg'])]['salary_year_avg']
```

```python
df_salary = df[pd.notna(df['salary_year_avg'])]
def projected_salary(salary):
    return salary * 1.03

df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)

df_salary[['salary_year_inflated', 'salary_year_avg']]
```

```python
df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)

df_salary[['salary_year_inflated', 'salary_year_avg']]  
```

```python
def projected_salary(row):
    if "Senior" in row['job_title']:
        return row['salary_year_avg'] * 1.05
    else:
        return row['salary_year_avg'] * 1.03
df_salary['salary_year_inflated'] = df_salary.apply(projected_salary, axis=1)

df_salary[['salary_year_inflated', 'salary_year_avg', 'job_title']]
```
