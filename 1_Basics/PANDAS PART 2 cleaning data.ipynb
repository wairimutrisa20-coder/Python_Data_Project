```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

```python
df.info()
```

```python
df.job_posted_date[0]
```

```python
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)   #changing datetime from str to datetime
```

```python
df['job_posted_month'] = df.job_posted_date.dt.month
```

```python
df.sort_values('job_posted_date')  # if you want to sort by something
```

```python
df.sort_values('job_posted_date', inplace=True)   #if you want the sort to be permanet onthe date set
```

```python
df.drop(labels='salary_hour_avg', axis=1,inplace=True) # if you want to drop a column and if you want permanently add ,inplace=True
```

```python
df.dropna(subset=['salary_year_avg'], inplace=True)    #drop rows with none values
```
